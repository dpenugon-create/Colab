Function 1: add_to_all_values
Write a function that takes two parameters, a dictionary of string:int pairs and another int. Given the parameter dictionary, your function should add the int parameter to every int value in the dictionary. Return the resulting dictionary.
add_to_all_values({"A": 1, "B": 2}, 2) --> {"A": 3, "B": 4}
add_to_all_values({"A": 1, "B": 2, "C": 3}, 1) --> {"A": 2, "B": 3, "C": 4}



```python
def add_to_all_values(dict1, int):
  for i in dict1:
    dict1[i] += int
  print(dict1)

add_to_all_values({'A':1, 'B':2,}, 2)
```

Function 2: get_value_sum
Write a function that takes one parameter, a dictionary of string:int pairs. The function should calculate and return the sum of all of the values in the dictionary.
get_value_sum({"dog": 2, "cat": 4}) --> 6
get_value_sum({"dog": 1, "cat": 1, "frog": 1}) --> 3


```python
def get_value_sum(dict2):
  x = 0
  for i in dict2:
    x += dict2[i]
  print(x)
get_value_sum({'dog':2, 'cat':4})
```

Function 3: append_to_values
Write a function that takes two parameters, a dictionary of string:list pairs and another int. The function should append the given int to each value list only if the int isn’t already in the value list. If the int is already in a value list, do nothing to that list.
append_to_values({"A": [], "B":[1,2]}, 5) --> {"A": [5], "B":[1,2,5]}
append_to_values({"A": [1,2,3], "B":[4,5,6]}, 2) --> {"A": [1,2,3], "B":[4,5,6,2]}


```python
def append_to_values(dict3, int3):
  for i in dict3:
    if int3 not in dict3[i]:
      dict3[i].append(int3)
  print(dict3)

append_to_values({'A':[], 'B':[1, 2]}, 5)
```

Function 4: convert_to_dictionary
Write a function that takes one specially formatted list as a parameter. The list will be formatted so that the elements in it alternate between string, int, string, int, string, int, etc. Your function should convert the list into a dictionary where each string is a key and the directly following int is it’s value. You can assume there are no duplicate strings.
convert_to_dictionary(["Alice", 10, "Bob", 20, "Eve", 30]) --> {"Alice": 10, "Bob": 20, "Eve": 30}
convert_to_dictionary(["Dog", 1, "Frog", 1]) --> {"Dog": 1, "Frog": 1}


```python
def convert_to_dictionary(lst):
  newdict = {}
  for i in range(0, len(lst), 2):
    newdict[lst[i]] = lst[i + 1]
  print(newdict)

convert_to_dictionary(['Alice', 10, 'Bob', 20, 'Eve', 30])

```
