Grocery Items Dictionary Practice

For this assignment, you’ll work on making a program that allows you to maintain a grocery list, including adding and deleting items.

Consider that we want to make a dictionary that represents our shopping list for groceries including:

4 apples
2 onions
1 loaf of bread
1 carton of eggs
2 potatoes
1 gallon milk

#Part One: Creating and Editing a Dictionary
Think about keys and values. To make our grocery dictionary for a shopping list, look at the elements above.
Which part should be the key? Which should be the value? Why?


1. Write code to create a dictionary that represents the shopping list above and stores it in a variable.

```python
grocerydict = {'apples':4, 'onions':1, 'bread':1, 'eggs':1, "potatoes":1, 'milk':1}
print(grocerydict)
```

2. Add the item 5 bananas to the dictionary, and print the list out

```python
grocerydict['banana'] = 5
print(grocerydict)
```

3. Write code for a loop that will go through and print out all elements in the dictionary in the format:
"You need ____ _________."
where the first blank is the number and the second is the item.
For example, the first entry would print "You need 4 apples."

```python
for x, y in grocerydict.items():
  print(f'You need {y} {x}')
```

4. Write code that allows the user to look up how many of a certain element they need.
For example, if they ask about “apples” it should say “You need 4 apples” but if you ask about “pears” it should say something like “Pears are not on your list”.

```python
request = input('Which item are you looking for ')
if request in grocerydict:
  print(f'You need {grocerydict.get(request)} {request}')
else:
  print(f'{request} are not on your list')
```

5. Write code to allow a user to add onto their list.
* You should first ask the user for the item they want to add.
At this point, there are two scenarios:
* Scenario 1: If that item is not yet in the dictionary, ask the user for how many of that item they need. Add this new key:value pair to the dictionary.
* For example, a user might say they wanted to add “oranges” and then said they wanted 5 of them, so add 5 oranges to the list.
* Scenario 2: If that item is already in the dictionary, you should ask them how many more of of that item they want. Update the dictionary to add that many of the item.
* For example, a user might say they want to update “apples”. Since “apples” is in their grocery dictionary, you’d ask the user how many more apples they want. The user would say a number, maybe 6, and then you’d change the apples entry to be 10 apples (4 initial + 6 added here).


```python
additem = input('What item do you want to add ')
if additem not in grocerydict:
  numitem = int(input(f'How many {additem} do you need '))
  grocerydict[additem] = numitem
else:
  numitem = int(input(f'How much more {additem} do you need '))
  grocerydict[additem] += numitem
print(grocerydict)
```
